# Simple Diffusion Model for Hill Slope Processes

This notebook demonstrates a one-dimensional hillslope diffusion model:

`dh/dt = diffusivity * d²h/dx²`

It uses explicit finite differences and no-flux boundary conditions.

```python
import matplotlib.pyplot as plt
import numpy as np

from diffusion_model import run_diffusion
```

```python
x = np.linspace(0, 1000, 101)
dx = x[1] - x[0]
initial_profile = 100 - 0.02 * (x - 500) ** 2

diffusivity = 0.01
dt = 1
num_steps = 100

final_profile = run_diffusion(
    initial_profile,
    diffusivity=diffusivity,
    dx=dx,
    dt=dt,
    num_steps=num_steps,
)
```

```python
plt.plot(x, initial_profile, label="Initial")
plt.plot(x, final_profile, label="After diffusion")
plt.xlabel("Distance (m)")
plt.ylabel("Elevation (m)")
plt.legend()
plt.title("Simple hillslope diffusion")
plt.show()
```
